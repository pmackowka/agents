A następnie:
1. Kliknij w miejsce, gdzie widnieje "Select Kernel" w prawym górnym rogu, i wybierz opcję `.venv (Python 3.12.12)` lub podobną - powinna być pierwszym albo najbardziej widocznym wyborem. Może być konieczne wcześniejsze wybranie "Python Environments".

1. Na Macu: z menu Cursor wybierz Settings >> VS Code Settings (UWAGA: koniecznie wybierz `VSCode Settings`, nie `Cursor Settings`);

2. W polu wyszukiwania Settings wpisz "venv"  
3. W polu "Path to folder with a list of Virtual Environments" wpisz ścieżkę do katalogu głównego projektu, np. C:\Users\username\projects\agents (na Windowsie) albo /Users/username/projects/agents (na Macu lub Linuksie).

```python
# A teraz - najważniejszy import
# Jeśli dostaniesz import error - zobacz Q5 i Q25 w FAQ na https://edwarddonner.com/avatar
# Nawet dla innych dostawców LLM, jak Gemini, nadal używasz tego importu z OpenAI - zobacz Guide 9, dlaczego

from openai import OpenAI
```

```python
# A teraz utworzymy instancję klasy OpenAI
# Jeśli nie jesteś pewien, co znaczy utworzyć instancję klasy - zobacz folder guides (guide 6)!
# Jeśli dostaniesz NameError - zobacz folder guides (guide 6), żeby dowiedzieć się o NameError - zawsze da się to od razu naprawić
# Jeśli nie używasz OpenAI, wystarczy to lekko zmodyfikować - dokładne instrukcje są w przewodniku po API AI (guide 9)

openai = OpenAI()
```

```python
# Utwórz listę wiadomości w znajomym formacie OpenAI

messages = [{"role": "user", "content": "Tell me a fun fact"}]
```

```python
messages
```

```python
# A teraz wywołaj to! Jeśli są problemy, zobacz przewodnik rozwiązywania problemów
# To używa GPT 5.4 nano, niesamowicie taniego modelu
# Przewodnik po API (guide 9) ma dokładne instrukcje, jak używać jeszcze tańszych albo darmowych alternatyw dla OpenAI
# Jeśli dostaniesz NameError, zobacz folder guides (guide 6), żeby dowiedzieć się o NameError - zawsze da się to od razu naprawić

response = openai.chat.completions.create(model="gpt-5.4-nano", messages=messages)
print(response.choices[0].message.content)
```

```python
# A teraz - poprośmy o pytanie:

question = "Please propose a hard, challenging question to assess someone's IQ. Respond only with the question."
messages = [{"role": "user", "content": question}]
```

```python
# zapytaj o to - to używa GPT 5.4 mini, wciąż tanie, ale mocniejsze niż nano

response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages)
question = response.choices[0].message.content
print(question)
```

```python
# utwórz nową listę messages

messages = [{"role": "user", "content": question}]
messages
```

```python
# Poproś model o odpowiedź na trudne pytanie

response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages)
answer = response.choices[0].message.content
print(answer)
```
