```python
# Sprawdź klucz - w tej wersji używamy Anthropic (Claude), więc potrzebny jest klucz ANTHROPIC_API_KEY

import os
anthropic_api_key = os.getenv('ANTHROPIC_API_KEY')

if anthropic_api_key:
    print(f"Anthropic API Key exists and begins {anthropic_api_key[:8]}")
else:
    print("Anthropic API Key not set - please head to the troubleshooting guide in the setup folder")
    
```

```python
# A teraz - najważniejszy import
# Jeśli dostaniesz import error, upewnij się, że masz zainstalowany pakiet: uv add anthropic
# W tej wersji używamy natywnego SDK Anthropic, a nie kompatybilnego wrappera OpenAI - zobacz Guide 9, dlaczego to inne podejście niż dla Gemini/DeepSeek/Ollama

from anthropic import Anthropic
```

```python
# A teraz utworzymy instancję klasy Anthropic
# Jeśli nie jesteś pewien, co znaczy utworzyć instancję klasy - zobacz folder guides (guide 6)!
# Jeśli dostaniesz NameError - zobacz folder guides (guide 6), żeby dowiedzieć się o NameError - zawsze da się to od razu naprawić
# Klient automatycznie odczyta klucz z ANTHROPIC_API_KEY w .env - nie trzeba go podawać ręcznie

anthropic = Anthropic()
```

```python
# Utwórz listę wiadomości w formacie Messages API (ten sam kształt role/content co w OpenAI)

messages = [{"role": "user", "content": "Opowiedz mi jakąś ciekawostkę"}]
```

```python
messages
```

```python
# A teraz wywołaj to! Jeśli są problemy, zobacz przewodnik rozwiązywania problemów
# To używa Claude Haiku 4.5, szybkiego i taniego modelu
# Anthropic wymaga podania max_tokens - to twardy limit długości odpowiedzi, inaczej niż w OpenAI, gdzie jest opcjonalny
# Jeśli dostaniesz NameError, zobacz folder guides (guide 6), żeby dowiedzieć się o NameError - zawsze da się to od razu naprawić

response = anthropic.messages.create(model="claude-haiku-4-5", max_tokens=1024, messages=messages)
print(response.content[0].text)
```

```python
# A teraz - poprośmy o pytanie:

question = "Zaproponuj trudne, wymagające pytanie do oceny czyjegoś IQ. Odpowiedz tylko samym pytaniem."
messages = [{"role": "user", "content": question}]
```

```python
# zapytaj o to - to używa Claude Sonnet 5, wciąż niedrogi, ale mocniejszy niż Haiku

response = anthropic.messages.create(model="claude-sonnet-5", max_tokens=1024, messages=messages)
question = response.content[0].text
print(question)
```

```python
# utwórz nową listę messages

messages = [{"role": "user", "content": question}]
messages
```

```python
# Poproś model o odpowiedź na trudne pytanie

response = anthropic.messages.create(model="claude-sonnet-5", max_tokens=1024, messages=messages)
answer = response.content[0].text
print(answer)
```
