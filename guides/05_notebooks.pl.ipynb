## Szerszy przewodnik po Notatnikach z przykładami

Zrób to teraz dla komórki poniżej:

## Gratulacje!

Teraz uruchom następną komórkę, która ustawia wartość, a potem kolejne komórki, żeby wypisać tę wartość

````python
# Ustaw wartość zmiennej

favorite_fruit = "bananas"
````

## Teraz wróć i uruchom ponownie komórkę z instrukcją print, dwie komórki wstecz

Widzisz, jak wypisuje coś innego, mimo że favorite_fruit zostało zmienione dalej w notatniku? 

Kolejność, w jakiej kod pojawia się w notatniku, nie ma znaczenia. Liczy się kolejność, w jakiej kod jest **wykonywany**. Za tym notatnikiem stoi proces Pythona, w którym zmieniane są zmienne.

To łapie niektóre osoby na początku pracy z notatnikami.

````python
# Potem uruchom tę komórkę dwa razy i sprawdź, czy rozumiesz, co się dzieje

print(f"My favorite fruit is {favorite_fruit}")

favorite_fruit = "apples"
````

# Wyjaśnienie 'kernela'

Za tym notatnikiem stoi proces Pythona, który wykonuje każdą komórkę, gdy ją uruchamiasz. Ten proces Pythona nazywa się Kernel. Każdy notatnik ma swój własny, osobny Kernel.

Możesz kliknąć przycisk powyżej "Restart Kernel".

Jeśli wtedy spróbujesz uruchomić następną komórkę, dostaniesz błąd, bo favorite_fruit nie jest już zdefiniowane. Będziesz musiał ponownie uruchomić komórki od góry notatnika. Wtedy następna komórka powinna zadziałać poprawnie.

````python
print(f"My favorite fruit is {favorite_fruit}")
````

# Drobne rzeczy, na które natrafiamy w tym kursie

To niekoniecznie funkcja notatników, ale fajny pakiet, warto o nim wiedzieć, bo jest przydatny w notatnikach.

Pakiet `tqdm` wypisze ładny pasek postępu, jeśli owiniesz nim dowolny iterowalny obiekt.

````python
# Oto trochę kodu bez paska postępu
# Zajmie to 10 sekund, podczas których zastanowisz się, co się dzieje..

import time

spams = ["spam"] * 1000

for spam in spams:
    time.sleep(0.01)
````

````python
# A teraz, z ładnym małym paskiem postępu:

import time
from tqdm import tqdm

spams = ["spam"] * 1000

for spam in tqdm(spams):
    time.sleep(0.01)
````

````python
# Zmieniając temat, oto przydatny sposób na wypisywanie outputu w markdown

from IPython.display import Markdown, display

display(Markdown("# This is a big heading!\n\n- And this is a bullet-point\n- So is this\n- Me, too!"))
````
