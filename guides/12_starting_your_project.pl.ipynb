## Jak zacząć pracę nad swoim pomysłem

Więc wpadłeś na pomysł na coś, co chciałbyś zbudować.

GRATULACJE! Ta najważniejsza część jest już zrobiona. Pomysł to wszystko!

Ale teraz nie jesteś pewien, od czego zacząć. Dostaję sporo pytań na ten temat.

Chcę Ci dać trzy kluczowe rady.


## Rada 1: Bądź Naukowcem

Bycie Inżynierem AI wymaga noszenia 2 kapeluszy: bycia Inżynierem Oprogramowania i Data Scientistem.

Na początku swojego projektu kluczowe jest, żeby zdjąć kapelusz Inżyniera Oprogramowania i mocno założyć kapelusz Data Science.

Wiele osób z backgroundem inżynierii oprogramowania (ja sam też!) ma z tym problem. Częste jest wpadanie w swoją strefę komfortu, zadawanie pytań o architekturę, design, potoki danych, skalowalność, wdrożenie.

To wszystko świetne pytania, ale nie są kluczowe na tym wczesnym etapie. To nie są trudne pytania. Nie zadecydują o powodzeniu Twojego pomysłu.

Do kluczowych pytań musisz być Naukowcem.

Zadaj sobie pytanie:  
1. Jak ocenisz sukces swojego modelu? Co będziesz mierzyć?  
2. Jakie dane masz, a jakich potrzebujesz?  
3. Jak zbudować małe prototypy, żeby przetestować różne techniki i zrozumieć wydajność modelu?

Zajmij się najpierw pytaniami naukowymi - co możesz osiągnąć z LLM-ami i jak.

I polecam zacząć w Notatniku, zanim zaczniesz pracować w modułach Pythona, żeby zmusić się do działania w tym naukowym nastawieniu.

To jest jeszcze bardziej kluczowe przy projektach z Agentami. Może kusić, żeby narysować duży diagram architektury Agentów - jak Twoi agenci będą współpracować, jak diagram techniczny. Ale to myślenie jak Inżynier Oprogramowania. Polecam podejść do tego inaczej; eksperymentuj z różnymi podejściami, badaj, co dobrze działa, testuj hipotezy i iteruj.

## Rada 2: Przyjmij R&D

OK, to jest dość podobne do Rady 1, ale warto powtórzyć!

Często dostaję pytania w stylu: "Ed, chcę zbudować takie rozwiązanie. Czy powinienem użyć modelu A, B, czy C? Czy powinienem użyć Agentów, RAG czy fine-tuningu?"

Moja odpowiedź jest niemal zawsze taka:  
- Powinieneś zrobić wszystko powyższe! Klucz to eksperymentowanie; (a) wymyśl kryteria oceny, (b) opracuj wyselekcjonowany zbiór danych, (c) przetestuj różne pomysły i zobacz, jak sobie radzą
- Często mam przeczucie, co zadziała najlepiej, ale moje przeczucie często jest błędne! Nie ufaj mi: sprawdź to sam  
- Po prostu nie ma zamiennika dla eksperymentowania.

## Rada 3: Marz wielko, ale zaczynaj MAŁO!

Kursanci też często wysyłają mi 200-300 linii kodu, mówiąc: "Ed, to nie działa jak trzeba, napraw to proszę". 😂

Jak mówię w przewodniku o debugowaniu - tak to nie działa!

Zaczynając projekty, kluczowe jest, żeby zaczynać małe i proste. Dopracuj każdy prompt; pracuj szczegółowo nad każdym krokiem. Upewnij się, że każde wywołanie LLM działa tak, jak chcesz, i iteruj nad danymi wejściowymi, aż odpowiedzi będą spójne i wiarygodne.

Jeśli zawsze pracujesz przyrostowo i z małymi, testowalnymi elementami składowymi, powinieneś mieć pełną jasność co do tego, co się dzieje.

I oczywiście uwielbiam pomagać ludziom z projektami i świetnie jest pomagać naprawiać problemy. Ale jeśli podchodzisz do swojego projektu przyrostowo, powinieneś zawsze mieć jasność co do tego, co dokładnie nie działa - i wtedy niezwykle efektywnie mogę Ci pomóc.
