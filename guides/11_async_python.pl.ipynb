### A teraz kilka przykładów:

````python
# Zdefiniujmy funkcję async

import asyncio

async def do_some_work():
    print("Starting work")
    await asyncio.sleep(1)
    print("Work complete")
````

````python
# Co to zrobi?

do_some_work()
````

````python
# OK spróbujmy jeszcze raz!

await do_some_work()
````

````python
# Co jest z tym nie tak?

async def do_a_lot_of_work():
    do_some_work()
    do_some_work()
    do_some_work()

await do_a_lot_of_work()
````

````python
# Ciekawe ostrzeżenie! Naprawmy to

async def do_a_lot_of_work():
    await do_some_work()
    await do_some_work()
    await do_some_work()

await do_a_lot_of_work()
````

````python
# A teraz zróbmy to równolegle
# Ważne jest, żeby zauważyć, że to nie jest "multi-threading" w sposobie, do jakiego możesz być przyzwyczajony
# Biblioteka asyncio działa na jednym wątku, ale używa pętli, żeby przełączać się między zadaniami, gdy jedno czeka

async def do_a_lot_of_work_in_parallel():
    await asyncio.gather(do_some_work(), do_some_work(), do_some_work())

await do_a_lot_of_work_in_parallel()
````
