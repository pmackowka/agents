# Podstawy techniczne

Kluczowe jest, żebyś czuł się komfortowo z podstawowymi koncepcjami technicznymi, z którymi pracujemy. To znacznie poprawi Twoje doświadczenie z całym kursem - może być bardzo frustrujące, jeśli nie jesteś pewien, co się dzieje.

Te przewodniki mają zbudować pewność siebie w podstawowych technologiach, z którymi pracujemy.

## Temat 1: ChatGPT kontra API OpenAI

### Jaka jest różnica między ChatGPT a API GPT, oba oferowane przez OpenAI?

#### ChatGPT to narzędzie dla użytkownika końcowego. To produkt typu Chat zaprojektowany dla konsumentów, którzy są użytkownikami AI.
- Ma darmowy plan, a także płatne plany subskrypcyjne z większą liczbą funkcji.
- Plany subskrypcyjne dają użytkownikowi niemal nieograniczony dostęp do korzystania z produktu Chat.

#### API to usługa dostarczana dla inżynierów AI - inżynierów oprogramowania i data scientistów - pracujących nad innymi produktami komercyjnymi.
- Pozwala technicznym osobom, takim jak Ty i ja, na dostęp do bazowych modeli (takich jak "GPT4.1" i "o3"), żebyśmy mogli budować własne produkty.
- Gdybyśmy chcieli, moglibyśmy zbudować własną wersję ChatGPT używając API, i pobierać opłaty od naszych użytkowników końcowych.
- Jak większość API, OpenAI pobiera niewielką opłatę na podstawie użycia API. Dla większości przykładów na kursie używających gpt-4o-mini, to rzędu $0,001 za wywołanie API.

### Płacę $20/miesiąc za ChatGPT - czemu muszę płacić więcej za API?

- Mam nadzieję, że teraz to jasne. API nie jest dla konsumentów; jest dla inżynierów, żeby budowali własne platformy, za które mogą pobierać opłaty.
- Gdybyś miał dostęp do API w ramach swojej subskrypcji, mógłbyś oferować narzędzia ChatGPT innym po niższej cenie i wypchnąć OpenAI z biznesu!
- Pamiętaj: każde wywołanie API może wymagać 10 000 000 000 000 operacji zmiennoprzecinkowych - te obliczenia zużywają prąd!

Zamiast wywoływać API, możesz uruchamiać modele open source lokalnie, ale zwykle mają 1000 razy mniej obliczeń — i mimo że to niewiele, to przetwarzanie i tak trafia na Twój rachunek za prąd..
