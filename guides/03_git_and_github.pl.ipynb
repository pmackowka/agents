### Wnoszenie swoich zmian do repo, żeby podzielić się swoim wkładem z innymi

Byłbym bardzo wdzięczny za włączenie Twojego wkładu. Daje to wartość wszystkim innym kursantom, a ja sam uwielbiam to oglądać! Jako dodatkowy bonus, dostajesz uznanie na Githubie jako współtwórca repo.

Oto ogólne kroki, jakie wiążą się z tworzeniem PR-a, i kluczowe instrukcje:  
https://edwarddonner.com/pr  

Sprawdź przed wysłaniem:  
1. Twój PR zawiera zmiany tylko w community_contributions (chyba że to omówiliśmy)  
2. Wszystkie outputy notatników są wyczyszczone  
3. Mniej niż 2000 linii kodu łącznie, i niezbyt dużo plików  
4. Nie dołączaj zbędnych plików testowych, przegadanego README ani .env.example, ani emoji, ani innych artefaktów LLM!

Bardzo dziękuję!
