# Witaj w Przewodnikach!

## Wprowadzenie

Zaprojektowałem ten kurs tak, żeby pasował do szerokiego zakresu doświadczeń. Dla tych z Was, którzy są w tym stosunkowo nowi, przygotowałem kilka technicznych briefingów, żeby zbudować Waszą wiedzę.

Są zaprojektowane do samodzielnej nauki; przechodzicie przez nie we własnym tempie, badając i eksperymentując.

Mocno skorzystałem z pomocy naszych przyjaciół AI, żeby napisać część przewodników, i starałem się je ułożyć tak, żeby były jak najbardziej przydatne i związane z kursem.

Kurs ma tylko jeden wymóg: sporo cierpliwości! Pamiętaj, że jednym z najlepszych sposobów nauki jest rozwiązywanie problemów - jeśli czujesz frustrację przy trudnej łamigłówce, pamiętaj, że właśnie tam dzieje się nauka! No i odezwij się, jeśli mogę pomóc.
