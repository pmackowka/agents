# Przewodnik przetrwania debugowania

## Klucz to konsekwentne odtworzenie swojego problemu w 5-10 liniach kodu

Trafienie na wyjątek czasem może być całkiem frustrujące, zwłaszcza jeśli nie wiesz, jak do tego podejść. Ale właśnie tam dzieje się najlepsza nauka! Docieranie do sedna trudnych problemów to świetny sposób na naukę, i może być całkiem satysfakcjonujące.

Ale jestem tu, żeby pomóc, jeśli nie dasz rady!

### Formuła na debugowanie: odtwórz w 10 liniach kodu

Świetne debugowanie to i sztuka, i nauka. Najlepszy sposób, żeby to opanować, to praktyka! Ale oto najważniejsze rzeczy:

1. Odtwórz  
Odtwórz swój problem, konsekwentnie

2. Uprość  
Zredukuj swój problem do najprostszego możliwego sposobu na jego odtworzenie. Nie powiem Ci, ilu ludzi wysłało mi 100+ linii kodu i powiedziało "Ed, to nie działa, napraw to proszę". Tak to nie działa! Pierwsza rzecz, jaką bym zrobił, to zredukowanie tego do sedna problemu - najlepiej w 10-20 liniach kodu albo mniej. W jednym przypadku pokazałem kursantowi, jak odtworzyć problem w 2 liniach kodu!

_Rzecz w tym, że **Tobie** jest dużo łatwiej to zrobić niż komukolwiek innemu._ Więc chociaż może to być żmudne, to prawie zawsze najlepszy sposób, żeby zacząć. Zredukuj do kilku linii kodu. A tak swoją drogą, gdy to zrobisz, często i tak sam odkryjesz problem!

3. Poproś o pomoc  
Gdy już odtworzyłeś problem w zaledwie kilku liniach kodu, jeśli nadal nie widzisz, co się dzieje - powinieneś poprosić o pomoc! Warto oczywiście zapytać ChatGPT i przyjaciół; daj im krótki przykład i pełny stack trace.

I odezwij się do mnie! Jestem tu, żeby uczynić Twoje doświadczenie nauki jak najlepszym, i jeśli utknąłeś, pomogę Ci się odblokować.

### Najlepszy sposób pracy ze mną, żeby dostać najszybsze poprawki...

1. Uprość problem najbardziej, jak się da, z łatwym sposobem na odtworzenie
2. Dołącz pełny stack trace i zrzut ekranu, nie zdjęcie (zobacz [ten przewodnik](https://chatgpt.com/share/681f691b-6644-8012-b07d-207c68f259d5), jeśli nie jesteś pewien, jak robić zrzuty ekranu)
3. Napisz do mnie na ed@edwarddonner.com. Albo najlepiej: jeśli używałeś kiedyś Google Colab, to naprawdę świetny sposób na podzielenie się problemem, bo będzie identyczny do odtworzenia dla mnie, i mogę go naprawić i podzielić się poprawką bezpośrednio z Tobą.

Nie mogę się doczekać, żeby Ci pomóc!
