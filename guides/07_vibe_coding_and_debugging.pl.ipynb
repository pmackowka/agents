# Vibe coding i debugowanie

"Vibe coding" to czuły termin na programowanie z pomocą LLM-ów. Jest szczególnie łatwe i wspaniałe z Cursorem! Ale są dobre techniki, które sprawiają, że działa to dobrze, i omawiam je na kursie. Oto kilka kluczowych wskazówek:

1. Jeśli promptujesz LLM taki jak ChatGPT czy Claude, żeby napisał kod, dołącz aktualną datę w swoim prompcie i poinstruuj LLM, żeby wygenerował kod używający najnowszych wersji API. To szczególnie ważne przy API OpenAI! ChatGPT często próbuje użyć przestarzałych wersji własnego API...

2. Też w swoich promptach, poproś LLM-y, żeby ich kod był jak najkrótszy i jak najprostszy. LLM-y zdają się uwielbiać dodawać mnóstwo zbędnych, dodatkowych sprawdzeń, które zaśmiecają kod i utrudniają znajdowanie problemów.

3. Zadaj to samo pytanie wielu LLM-om i wybierz odpowiedź, która jest najbardziej klarowna i najprostsza.

4. Podobnie: weź odpowiedź od 1 LLM-a i poproś inny LLM o zweryfikowanie jej poprawności i tego, czy mogłaby być prostsza albo jaśniejsza.

5. Poproś LLM o podanie 3 wariantów rozwiązania problemu i wybierz ten najbardziej klarowny.

## Vibe Coding dla większego projektu

Staraj się unikać sytuacji, w której LLM generuje 100-200 linii kodu albo więcej; będzie Ci bardzo trudno to zdebugować i zrozumieć, co idzie nie tak (chyba że już masz sporą wiedzę).

Zamiast tego: zacznij od rozbicia swojego problemu na małe, niezależnie testowalne kroki, każdy stosunkowo mały. Jeśli nie jesteś pewien, jak rozbić swój problem - to jest coś, o co możesz poprosić LLM!

Potem dla każdego z tych elementów składowych:
- Użyj powyższych wskazówek, żeby LLM zbudował kod
- Poproś też LLM o napisanie testów, żeby przetestować i zweryfikować kod
- Przetestuj to sam i upewnij się, że działa poprawnie

To pozwoli Ci z pewnością siebie zbudować większy projekt.
